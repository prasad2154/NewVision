```python
from sentence_transformers import SentenceTransformer, util

sentences = [
     "The car is very fast and has excellent performance.",
    "The vehicle is extremely quick and performs exceptionally well.",
    "I enjoy cooking Italian food on weekends."
]

model = SentenceTransformer("all-MiniLM-L6-v2")

embeddings = model.encode(sentences)

print("Comparing sentences:\n")

print(f"1: '{sentences[0]}'")
print(f"2: '{sentences[1]}'")
print(f"3: '{sentences[2]}'")

score_12 = util.cos_sim(embeddings[0], embeddings[1])
score_13 = util.cos_sim(embeddings[0], embeddings[2])

print(f"\n1 vs 2 (similar meaning): {score_12.item():.4f}")
print(f"1 vs 3 (different topic):  {score_13.item():.4f}")
```

```python
from openai import OpenAI
from sklearn.metrics.pairwise import cosine_similarity

client = OpenAI()

sentences = [
     "The car is very fast and has excellent performance.",
    "The vehicle is extremely quick and performs exceptionally well.",
    "I enjoy cooking Italian food on weekends."
]

# Generate embeddings
embeddings = []

for s in sentences:
    response = client.embeddings.create(
        model="text-embedding-3-small",
        input=s
    )

    embeddings.append(response.data[0].embedding)

# Compare
print("Comparing sentences:\n")

print(f"1: '{sentences[0]}'")
print(f"2: '{sentences[1]}'")
print(f"3: '{sentences[2]}'")

print(
    f"\n1 vs 2 (similar meaning): "
    f"{cosine_similarity([embeddings[0]], [embeddings[1]])[0][0]:.4f}"
)

print(
    f"1 vs 3 (different topic):  "
    f"{cosine_similarity([embeddings[0]], [embeddings[2]])[0][0]:.4f}"
)
```
